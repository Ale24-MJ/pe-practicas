## Proyecto de forma local

Si estas trabajando el proyecto de forma local, primero debes ubicar tu base de datos en una ruta conocida, de preferencia en la misma carpeta que este notebook.

```python
import pandas as pd

# Cargamos el archivo CSV, usando la ruta completa donde se encuentra el archivo
datos = pd.read_csv('./Social_media_impact_on_life.csv')

# Mostrar los primeros registros
display(datos.head())

# Consultamos las dimensiones del conjunto de datos
print("Numero de registros: ", datos.shape[0])
print("Numero de variables: ", datos.shape[1])

# Mostramos los nombres de las columnas
print(datos.columns)
```
