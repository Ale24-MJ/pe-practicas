## Ejercicio práctico 1.1
Ahora comprobaremos mediante una simulación cómo los resultados experimentales pueden aproximarse a una probabilidad teórica.

Consideremos nuevamente un dado justo.El evento de interés será:  
$A = {\text{obtener un número par}}$  

La probabilidad teórica dice:  
$P(A) = \frac{3}{6} = \frac{1}{2} = 50%$

Ahora utilzaremos Python y la biblioteca `random` para simular 10, 100 y 1000 lanzamientos de datos.

## Intrucciones
1. Genera la cantidad indicada de resultados aleatorios entre 1 y 6.
1. Cuenta cuántos resultados pertenecen al evento $A$.
1. Calcula la frecuencia relativa: $P_{experimental}(A)$ = $\frac{\text{número de resultados favorables}}{\text{número total de lanzamientos}}$
1. Compara el resuyltado expemiental con: $P_{teórica}(A) = 0.5$
1. Calcula la diferencia entre ambos valores.

```python
import random

lanzamientos = 1000
favorables = 0

for _ in range(lanzamientos):
    resultados = random.randint(1, 6)

    # Determina si el resultado pertenece al evento A
    if(resultados%2 == 0):
        # Actualiza el contador correspondiente
        favorables += 1 

probabilidad_experimental = favorables / lanzamientos

print("Lanzamientos: ", lanzamientos)
print("Resultados favorables: ", favorables)
print("Probabilidad Experimental: ", probabilidad_experimental)
print("Probabilidad Teórica: ", 0.5)
```
