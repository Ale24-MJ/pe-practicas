# Simulación

Evaluaremos 1000 lanzamientos de dado, queremos evaluar la probabilidad de que salga un numero par, si queremos comprobar la probabilidad teórica con la probabilidad experimental necesitamos hacer lo siguiente:

```python
import random

lanzamientos = 10 # Esta variable nos indicará el numero de lanzamientos que realizaremos. 
pares = 0 # En este contador acumularemos el numero de numeros pares que aparezcan en el dado.

for _ in range(lanzamientos):
  resultado = random.randint(1, 6) # Esta linea genera un entero pseudoaleatorio entre 1 y 6

  if (resultado % 2 == 0):
    pares += 1 # Aqui incrementamos el contador cuando sale un numero par en el resultado del lanzamiento

# Calculamos la probabilidad experimental
probabilidad_experimental = pares / lanzamientos

print("Lanzamientos: ", lanzamientos)
print("Resultados pares: ", pares)
print("Probabilidad experimental: ", probabilidad_experimental)
print("Probabilidad teorica: ", 0.5)
```

Ahora verifiquemos el cálculo utilizando conjuntos.

```python
espacio = {1, 2, 3, 4, 5, 6} # Espacio muestral

A = {2, 4, 6}
B = {4, 5, 6}

interseccion = A & B

prob_A_interseccion_B = len(interseccion) / len(espacio)
prob_B = len(B) / len(espacio)

prob_A_dado_B = prob_A_interseccion_B / prob_B

print("A ∩ B = ", interseccion)
print("P(A ∩ B) = ", prob_A_interseccion_B)
print("P(B) = ", prob_B)
print("P(A | B) = ", prob_A_dado_B)
```

## Independencia
Ahora analizaremos una situación diferente, consideremos dos experimentos independientes:
1. Lanzar una moneda
1. Lanzar un dado

Definimos los siguientes eventos:  
$A = \{\text{obtener águla en la moneda}\}$  
$B = \{\text{obtener un número par en el dado}\}$

La ocurrencia de A no modifica las posibilidades del dado, por lo tanto esperamos que:  
$P(B|A) = P(B)$

Y de forma equivalente  
$P(A \cap B) = P(A) P(B)$

Utilizaremos `random` para simular 10000 repeticiones del experimento.

En cada repetición debemos:
1. Lanzar una moneda.
1. Lanzar un dado.
1. Determinar si ocurrió A.
1. Determinar si ocurrió B.
1. Determinar si ocurrieron A y B.

```python
import random

repeticiones = 10000

evento_A = 0
evento_B = 0
evento_A_y_B = 0

for _ in range(repeticiones):
  moneda = random.randint(0,1) # 1
  dado = random.randint(1,6) # 2

  # 3,4,5 Determina si ocurre el evento A o el evento B o ambos.
  if (moneda == 0): evento_A += 1
  if (dado % 2 == 0): evento_B += 1
  if (moneda == 0 and dado % 2 == 0): evento_A_y_B += 1


# Calcula las probabilidad experimental del evento A, evento B e intersección del evento A y B.
prob_A = evento_A / repeticiones
prob_B = evento_B / repeticiones
prob_A_y_B = evento_A_y_B / repeticiones

print("P(A) = ",prob_A)
print("P(B) = ",prob_B)
print("P(A ∩ B) = ",prob_A_y_B)
```
